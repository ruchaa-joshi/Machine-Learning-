```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression, LogisticRegression
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
from sklearn.metrics import confusion_matrix, classification_report
```

```python
np.random.seed(0)

X_linear = 2 * np.random.rand(100, 1)

y_linear = 4 + 3 * X_linear + np.random.randn(100, 1)

print("First 5 X values:")
print(X_linear[:5])

print("\nFirst 5 y values:")
print(y_linear[:5])
```

```python
plt.scatter(X_linear, y_linear)

plt.xlabel("X")
plt.ylabel("Y")
plt.title("Linear Regression Data")

plt.show()
```

```python
X_train_linear, X_test_linear, y_train_linear, y_test_linear = train_test_split(
    X_linear,
    y_linear,
    test_size=0.2,
    random_state=42
)

print("Training samples:", len(X_train_linear))
print("Testing samples:", len(X_test_linear))
```

```python
linear_model = LinearRegression()

linear_model.fit(X_train_linear, y_train_linear)

print("Intercept:", linear_model.intercept_)
print("Coefficient:", linear_model.coef_)
```

```python
y_pred_linear = linear_model.predict(X_test_linear)

print("Actual values:")
print(y_test_linear[:10])

print("\nPredicted values:")
print(y_pred_linear[:10])
```

```python
mse = mean_squared_error(y_test_linear, y_pred_linear)
rmse = np.sqrt(mse)
mae = mean_absolute_error(y_test_linear, y_pred_linear)
r2 = r2_score(y_test_linear, y_pred_linear)

print("Linear Regression Evaluation")
print("----------------------------")
print("MSE  :", round(mse, 2))
print("RMSE :", round(rmse, 2))
print("MAE  :", round(mae, 2))
print("R²   :", round(r2, 2))
```

```python
plt.scatter(X_test_linear, y_test_linear, label="Actual")
plt.plot(X_test_linear, y_pred_linear, label="Predicted")

plt.xlabel("X")
plt.ylabel("Y")
plt.title("Linear Regression")

plt.legend()
plt.show()
```

```python
X_logistic = np.random.randn(100, 1)

y_logistic = (X_logistic > 0).astype(int).ravel()

print("First 10 X values:")
print(X_logistic[:10])

print("\nFirst 10 class values:")
print(y_logistic[:10])
```

```python
plt.scatter(X_logistic, y_logistic)

plt.xlabel("X")
plt.ylabel("Class")
plt.title("Logistic Regression Data")

plt.yticks([0, 1])

plt.show()
```

```python
X_train_logistic, X_test_logistic, y_train_logistic, y_test_logistic = train_test_split(
    X_logistic,
    y_logistic,
    test_size=0.2,
    random_state=42
)

print("Training samples:", len(X_train_logistic))
print("Testing samples:", len(X_test_logistic))
```

```python
logistic_model = LogisticRegression()

logistic_model.fit(X_train_logistic, y_train_logistic)

print("Intercept:", logistic_model.intercept_)
print("Coefficient:", logistic_model.coef_)
```

```python
y_pred_logistic = logistic_model.predict(X_test_logistic)

print("Actual values:")
print(y_test_logistic)

print("\nPredicted values:")
print(y_pred_logistic)
```

```python
y_probability = logistic_model.predict_proba(X_test_logistic)

print("Prediction probabilities:")
print(y_probability[:10])
```

```python
accuracy = accuracy_score(y_test_logistic, y_pred_logistic)
precision = precision_score(y_test_logistic, y_pred_logistic)
recall = recall_score(y_test_logistic, y_pred_logistic)
f1 = f1_score(y_test_logistic, y_pred_logistic)

print("Logistic Regression Evaluation")
print("------------------------------")
print("Accuracy :", round(accuracy, 2))
print("Precision:", round(precision, 2))
print("Recall   :", round(recall, 2))
print("F1 Score :", round(f1, 2))
```

```python
cm = confusion_matrix(y_test_logistic, y_pred_logistic)

print("Confusion Matrix:")
print(cm)
```

```python
print(classification_report(y_test_logistic, y_pred_logistic))
```

```python
plt.scatter(
    X_test_logistic,
    y_test_logistic,
    label="Actual"
)

plt.scatter(
    X_test_logistic,
    y_pred_logistic,
    marker="x",
    label="Predicted"
)

plt.xlabel("X")
plt.ylabel("Class")
plt.title("Logistic Regression")

plt.yticks([0, 1])
plt.legend()

plt.show()
```

```python
print("===== FINAL RESULTS =====")

print("\nLinear Regression")
print("MSE      :", round(mse, 2))
print("RMSE     :", round(rmse, 2))
print("MAE      :", round(mae, 2))
print("R² Score :", round(r2, 2))

print("\nLogistic Regression")
print("Accuracy :", round(accuracy, 2))
print("Precision:", round(precision, 2))
print("Recall   :", round(recall, 2))
print("F1 Score :", round(f1, 2))
```
