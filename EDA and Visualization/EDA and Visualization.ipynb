```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
titanic = pd.read_csv('titanic.csv')
```

```python
# Display basic information about columns, data types and missing values
titanic.info()
```

```python
# Display the first 5 rows
titanic.head()
```

```python
# Display the last 5 rows
titanic.tail()
```

```python
titanic.info()

print("\nFirst 5 rows:")
display(titanic.head())

print("\nLast 5 rows:")
display(titanic.tail())
```

```python
# Statistical summary of numerical columns
titanic.describe()
```

```python
# Transpose the summary statistics for easier reading
titanic.describe().transpose()
```

```python
# Count how many passengers survived and died
print(titanic['survived'].value_counts())

# 0 = Did not survive
# 1 = Survived
```

```python
# Visualize the number of survivors and non-survivors
sns.countplot(data=titanic, x='survived')

plt.xlabel('Survived')
plt.ylabel('Number of Passengers')
plt.title('Number of Survivors and Non-Survivors')

plt.show()
```

```python
# Count passengers in each passenger class
print(titanic['pclass'].value_counts())
```

```python
# 'pclass' is numerical while 'class' contains text labels

print("Pclass:")
print(titanic['pclass'].value_counts())

print("\nClass:")
print(titanic['class'].value_counts())
```

```python
# Display pclass and class side by side
p_class = titanic[['pclass', 'class']]

p_class.head()
```

```python
# Count passengers according to gender
print(titanic['sex'].value_counts())
```

```python
# Visualize the number of male and female passengers
sns.countplot(data=titanic, x='sex')

plt.xlabel('Gender')
plt.ylabel('Number of Passengers')
plt.title('Number of Male and Female Passengers')

plt.show()
```

```python
# Select passengers whose age is less than 20
less_than_20 = titanic[titanic['age'] < 20]

# Display the first few passengers
less_than_20.head()
```

```python
# Count the number of passengers younger than 20
len(less_than_20)
```

```python
# Count the different passenger categories
titanic['who'].value_counts()
```

```python
# Display passenger categories using a pie chart
titanic['who'].value_counts().plot(
    kind='pie',
    autopct='%1.1f%%'
)

plt.title('Passenger Categories')
plt.ylabel('')

plt.show()
```

```python
# Find the number of unique cities/towns where passengers boarded
print("Number of unique embark towns:")
print(titanic['embark_town'].nunique())
```

```python
# Count passengers from each embarkation town
print(titanic['embark_town'].value_counts())
```

```python
# Plot the number of passengers from each embarkation town
titanic['embark_town'].value_counts().plot(kind='bar')

plt.xlabel('Embark Town')
plt.ylabel('Number of Passengers')
plt.title('Passengers by Embarkation Town')

plt.show()
```

```python
# Count missing values in every column
titanic.isnull().sum()
```

```python
# Visualize missing values in the dataset
sns.heatmap(
    titanic.isnull(),
    yticklabels=False,
    cbar=False
)

plt.title('Missing Data in Titanic Dataset')
plt.show()
```

```python
# Compare survival with gender
sns.countplot(
    data=titanic,
    x='survived',
    hue='sex',
    palette='autumn'
)

plt.xlabel('Survived')
plt.ylabel('Number of Passengers')
plt.title('Survival According to Gender')

plt.show()
```

```python
# Compare survival with passenger class
sns.countplot(
    data=titanic,
    x='survived',
    hue='class',
    palette='viridis'
)

plt.xlabel('Survived')
plt.ylabel('Number of Passengers')
plt.title('Survival According to Passenger Class')

plt.show()
```

```python
# Display the distribution of passenger ages
titanic['age'].plot(
    kind='hist',
    bins=30
)

plt.xlabel('Age')
plt.ylabel('Number of Passengers')
plt.title('Age Distribution of Titanic Passengers')

plt.show()
```

```python
# Scatter plot showing relationship between age and fare
sns.scatterplot(
    data=titanic,
    x='age',
    y='fare',
    hue='class'
)

plt.title('Age vs Fare by Passenger Class')
plt.show()
```

```python
# Scatter plot showing relationship between age and fare
sns.scatterplot(
    data=titanic,
    x='age',
    y='fare',
    hue='sex'
)

plt.title('Age vs Fare by Gender')
plt.show()
```
